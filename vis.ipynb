```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
"""
count battery id's (categories)
"""

fig, ax = plt.subplots()
df = pd.read_csv("battery_observations.csv", index_col=0)
print(df)
cats = list(set(df.index))
print('num cats:', len(cats))
counts = df.index.value_counts()
ax.bar(counts.index, counts.values)
plt.show()

```

```python
"""
count degradation_status categories
"""

fig, ax = plt.subplots()
df = pd.read_csv("battery_observations.csv", index_col=0)
cats = list(set(df["degradation_status"]))
print('num cats:', len(cats))
counts = df["degradation_status"].value_counts()
ax.bar(counts.index, counts.values)
plt.show()

```

```python
"""
count chemistries, usage_strategy, climate zone
"""

# 1 row, 2 columns
fig, ax = plt.subplots(2, 2, figsize=(10, 4))
metadata = pd.read_csv("battery_metadata.csv", index_col=0)
print(metadata)
cats = list(set(metadata["chemistry"]))
print('num cats:', len(cats))
counts = metadata["chemistry"].value_counts()
ax[0,0].bar(counts.index, counts.values)
counts = metadata["climate_zone"].value_counts()
ax[1,0].bar(counts.index, counts.values)
counts = metadata["usage_strategy"].value_counts()
ax[0,1].bar(counts.index, counts.values)
fig.tight_layout()
plt.show()

```
